## Pertanyaan Bisnis Utama:
1. **Faktor Lembur:** Apakah karyawan yang sering kerja lembur (OverTime) memiliki kecenderungan lebih tinggi untuk resign (Attrition)?
2. **Keseimbangan Gaji:** Apakah tingkat pendapatan bulanan (Monthly Income) mempengaruhi keputusan karyawan untuk meninggalkan perusahaan?
3. **Kepuasan & Lingkungan:** Bagaimana hubungan antara kepuasan lingkungan kerja (Environment Satisfaction) terhadap loyalitas karyawan?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data (Nama file biasanya 'WA_Fn-UseC_-HR-Employee-Attrition.csv')
df = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')

# Cek kolom yang tersedia
print(df.columns)
df.head(10)
```

## 1. Analisis Lembur vs Attrition
Di sini kita akan menjawab: **"Apakah karyawan yang sering kerja lembur (OverTime) memiliki kecenderungan lebih tinggi untuk resign (Attrition)?"**

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='OverTime', hue='Attrition', palette='viridis')
plt.title('Dampak Lembur terhadap Attrition Karyawan', fontsize=14)
plt.xlabel('Bekerja Lembur (OverTime)', fontsize=12)
plt.ylabel('Jumlah Karyawan', fontsize=12)
plt.legend(title='Attrition', loc='upper right')
plt.show()
```

## 2. Analisis Pendapatan
Di sini kita akan menjawab: **"Apakah tingkat pendapatan bulanan (Monthly Income) mempengaruhi keputusan karyawan untuk meninggalkan perusahaan?"**

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Attrition', y='MonthlyIncome', palette='magma')
plt.title('Perbandingan Pendapatan Bulanan: Bertahan vs Resign', fontsize=14)
plt.xlabel('Status Attrition', fontsize=12)
plt.ylabel('Monthly Income ($)', fontsize=12)
plt.show()
```

## 3. Analisis Kepuasan Lingkungan
Di sini kita akan menjawab: **"Bagaimana hubungan antara kepuasan lingkungan kerja (Environment Satisfaction) terhadap loyalitas karyawan?"**

```python
plt.figure(figsize=(10, 6))
# 1: Low, 2: Medium, 3: High, 4: Very High
sns.barplot(data=df, x='EnvironmentSatisfaction', y='MonthlyIncome', hue='Attrition', ci=None, palette='coolwarm')
plt.title('Analisis Attrition Berdasarkan Kepuasan Lingkungan & Gaji', fontsize=14)
plt.xlabel('Tingkat Kepuasan Lingkungan (1-4)', fontsize=12)
plt.show()
```

## Kesimpulan
### 1. Dampak Lembur (OverTime) terhadap Retensi
- **Insight:** Karyawan yang melakukan kerja lembur (OverTime) menunjukkan tingkat pengunduran diri (Attrition) yang jauh lebih tinggi secara proporsional dibandingkan mereka yang tidak lembur. Meskipun jumlah karyawan yang tidak lembur lebih banyak, rasio karyawan yang 'Resign' pada kelompok lembur sangat mencolok. Ini menunjukkan bahwa beban kerja berlebih menjadi faktor penentu kelelahan karyawan (*burnout*).
- **Rekomendasi:** Perusahaan perlu meninjau kembali distribusi beban kerja dan kebijakan kompensasi lembur untuk mengurangi tingkat pengunduran diri pada posisi-posisi kritis.
### 2. Analisis Pendapatan Bulanan (Monthly Income)
- **Insight:** Terdapat perbedaan signifikan antara pendapatan karyawan yang bertahan dibandingkan yang keluar. Median gaji karyawan yang resign berada jauh di bawah median gaji karyawan yang bertahan. Faktor finansial memegang peranan penting. Karyawan dengan pendapatan rendah memiliki tingkat kerentanan yang lebih tinggi untuk mencari peluang di luar perusahaan.
- **Rekomendasi:** Perusahaan perlu meninjau kembali standar gaji untuk level staf atau posisi entry-level, karena kelompok inilah yang paling banyak mengalami attrition.
### 3. Hubungan Kepuasan Lingkungan, Gaji, dan Attrition
- **Insight:** Karyawan yang resign di tingkat kepuasan lingkungan 3 dan 4 memiliki kesenjangan gaji yang sangat besar dibandingkan rekan mereka yang bertahan. Ini membuktikan bahwa faktor kompensasi tetap menjadi pertimbangan utama meskipun budaya kantor sudah baik.
- **Rekomendasi:** Perusahaan tidak boleh hanya fokus pada fasilitas kantor, tetapi juga harus memastikan struktur gaji tetap kompetitif untuk mencegah 'pembajakan' karyawan berkualitas oleh kompetitor.
